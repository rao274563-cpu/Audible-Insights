```python
df = pd.read_csv("../data/raw/Audible_Catlog.csv")
```

```python
common_books = set(df["Book Name"]) & set(df_advance["Book Name"])

print("Common books:", len(common_books))
print("Only in Dataset 1:", df["Book Name"].nunique() - len(common_books))
print("Only in Dataset 2:", df_advance["Book Name"].nunique() - len(common_books))
```

```python
only_dataset1 = df[
    ~df["Book Name"].isin(df_advance["Book Name"])
]

only_dataset1[["Book Name", "Author", "Rating", "Price"]].head(10)
```

```python
only_dataset2 = df_advance[
    ~df_advance["Book Name"].isin(df["Book Name"])
]

only_dataset2[
    ["Book Name", "Author", "Rating", "Price", "Description"]].head(10)
```

```python
common_books_df = pd.DataFrame({
    "Book Name": list(common_books)
})

common_books_df["Dataset1_Authors"] = common_books_df["Book Name"].map(
    df.groupby("Book Name")["Author"].nunique()
)

common_books_df["Dataset2_Authors"] = common_books_df["Book Name"].map(
    df_advance.groupby("Book Name")["Author"].nunique()
)

common_books_df[
    (common_books_df["Dataset1_Authors"] > 1) |
    (common_books_df["Dataset2_Authors"] > 1)
].head(10)
```

```python
df[
    df["Book Name"].isin(
        common_books_df.loc[
            (common_books_df["Dataset1_Authors"] > 1) |
            (common_books_df["Dataset2_Authors"] > 1),
            "Book Name"
        ]
    )
][["Book Name", "Author"]].sort_values("Book Name")
```

```python
df_advance[
    df_advance["Book Name"].isin(
        common_books_df.loc[
            (common_books_df["Dataset1_Authors"] > 1) |
            (common_books_df["Dataset2_Authors"] > 1),
            "Book Name"
        ]
    )
][["Book Name", "Author"]].sort_values("Book Name")
```

```python
dataset1_pairs = set(zip(df["Book Name"], df["Author"]))
dataset2_pairs = set(zip(df_advance["Book Name"], df_advance["Author"]))

print("Matching book-author pairs:", len(dataset1_pairs & dataset2_pairs))
print("Dataset 1 exclusive pairs:", len(dataset1_pairs - dataset2_pairs))
print("Dataset 2 exclusive pairs:", len(dataset2_pairs - dataset1_pairs))
```

```python
comparison = df.merge(
    df_advance,
    on=["Book Name", "Author"],
    suffixes=("_dataset1", "_dataset2")
)

comparison[
    ["Book Name", "Author", "Rating_dataset1", "Rating_dataset2"]
].head(10)
```

```python
rating_differences = comparison[
    comparison["Rating_dataset1"] != comparison["Rating_dataset2"]
]

print("Rows with different ratings:", len(rating_differences))

rating_differences[
    ["Book Name", "Author", "Rating_dataset1", "Rating_dataset2"]
].head(10)
```

```python
review_differences = comparison[
    comparison["Number of Reviews_dataset1"] !=
    comparison["Number of Reviews_dataset2"]
]

print("Rows with different review counts:", len(review_differences))

review_differences[
    ["Book Name", "Author",
     "Number of Reviews_dataset1",
     "Number of Reviews_dataset2"]
].head(10)
```

```python
df[
    df.duplicated(["Book Name", "Author"], keep=False)
].sort_values(["Book Name", "Author"])[
    ["Book Name", "Author", "Rating", "Number of Reviews", "Price"]
].head(20)
```

```python
def inspect_duplicate_variations(data, name):
    varying_pairs = data.groupby(["Book Name", "Author"]).filter(
        lambda group: len(group.drop_duplicates()) > 1
    )

    print(f"\n{name}")
    print("Duplicate pairs with different records:",
          varying_pairs.groupby(["Book Name", "Author"]).ngroups)
    print("Total rows in these pairs:", len(varying_pairs))

    display(
        varying_pairs.sort_values(["Book Name", "Author"]).head(10)
    )


inspect_duplicate_variations(df, "Dataset 1")
inspect_duplicate_variations(df_advance, "Dataset 2")
```

```python
core_columns = ["Rating", "Number of Reviews", "Price"]

varying_core = df_advance.groupby(
    ["Book Name", "Author"]
).filter(
    lambda group: any(group[col].nunique() > 1 for col in core_columns)
)

print("Pairs with conflicting core values:",
      varying_core.groupby(["Book Name", "Author"]).ngroups)

display(
    varying_core.sort_values(["Book Name", "Author"])[
        ["Book Name", "Author", "Rating", "Number of Reviews", "Price"]
    ].head(20)
)
```

```python
core_columns = ["Rating", "Number of Reviews", "Price"]

varying_core_d1 = df.groupby(
    ["Book Name", "Author"]
).filter(
    lambda group: any(group[col].nunique() > 1 for col in core_columns)
)

print("Pairs with conflicting core values:",
      varying_core_d1.groupby(["Book Name", "Author"]).ngroups)

display(
    varying_core_d1.sort_values(["Book Name", "Author"])[
        ["Book Name", "Author", "Rating", "Number of Reviews", "Price"]
    ].head(20)
)
```
